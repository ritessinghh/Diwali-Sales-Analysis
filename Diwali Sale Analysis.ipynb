```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt # visualizing data
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv('Diwali Sales Data.csv', encoding = 'unicode_escape')
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1, inplace=True)
```

```python
df.info()
```

```python
pd.isnull(df).sum()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df.info()
```

```python
df['Amount'] = df['Amount'].astype('int')
```

```python
df['Amount'].dtypes
```

```python
df.describe()
```

```python
df[['Age', 'Orders', 'Amount']].describe()
```

```python
df.columns
```

```python
ax = sns.countplot(x='Gender' , data =df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Gender',y= 'Amount' ,data = sales_gen)
```

```python
sale_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x = 'Gender' , y = 'Amount', data = sale_gen)
```

```python
ax = sns.countplot(x='Age Group' , hue ='Gender',data = df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x = 'Age Group' , y = 'Amount', data = sale_age)
```

```python
State Analysis
```

```python
df.columns
```

```python
ax = sns.countplot(x='State' , data =df)
sns.set(rc={'figure.figsize': (25,4)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(8)
sns.set(rc={'figure.figsize':(18,6)})
sns.barplot(x = 'State' , y = 'Orders', data = sale_state)
```

```python
sale_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(8)
sns.set(rc={'figure.figsize':(18,6)})
sns.barplot(x = 'State' , y = 'Amount', data = sale_state)
```
